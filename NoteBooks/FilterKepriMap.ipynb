```python
import json
from pathlib import Path

# =========================================================
# FILE
# =========================================================
INPUT_FILE = Path("C:/Users/HP/PycharmProjects/DashboardPenyaluranBansos/Geo/gadm41_IDN_3.json")
OUTPUT_FILE = Path("C:/Users/HP/PycharmProjects/DashboardPenyaluranBansos/Geo/Map_Kecamatan_Kepri.geojson")

# =========================================================
# LOAD GADM
# =========================================================
with INPUT_FILE.open("r", encoding="utf-8") as f:
    data = json.load(f)

# =========================================================
# FILTER PROVINSI KEPULAUAN RIAU
# =========================================================
features_kepri = [
    feature
    for feature in data["features"]
    if feature.get("properties", {}).get("NAME_1") == "KepulauanRiau"
]

print("Jumlah kecamatan:", len(features_kepri))

# =========================================================
# TAMPILKAN KABUPATEN/KOTA
# =========================================================
kabupaten = sorted({
    feature["properties"].get("NAME_2")
    for feature in features_kepri
})

print("\nKabupaten/Kota:")
for nama in kabupaten:
    print("-", nama)

# =========================================================
# TAMBAHKAN FIELD YANG LEBIH MUDAH DIPAKAI DASHBOARD
# =========================================================
for feature in features_kepri:

    properties = feature["properties"]

    properties["PROVINSI"] = "KEPULAUAN RIAU"

    properties["KABUPATEN_KOTA"] = (
        properties.get("NAME_2", "")
        .upper()
    )

    properties["KECAMATAN"] = (
        properties.get("NAME_3", "")
        .upper()
    )

# =========================================================
# BUAT GEOJSON BARU
# =========================================================
output_data = {
    "type": "FeatureCollection",
    "features": features_kepri
}

with OUTPUT_FILE.open("w", encoding="utf-8") as f:
    json.dump(
        output_data,
        f,
        ensure_ascii=False,
        separators=(",", ":")
    )

print("\nBerhasil dibuat:")
print(OUTPUT_FILE)
```
